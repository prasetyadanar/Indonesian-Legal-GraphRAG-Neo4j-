## **Embedding**

```python
import os
from langchain_openai import OpenAIEmbeddings
from langchain_neo4j import Neo4jVector
from neo4j_graphrag.retrievers import HybridCypherRetriever, VectorCypherRetriever
from pprint import pprint
```

```python
embeddings = OpenAIEmbeddings(
    model="text-embedding-3-large",
    base_url=os.environ["SUMOPOD_URI"] + "/v1",
    api_key=os.environ["SUMOPOD_KEY"]
)
```

```python
# 'Embedded' ke pasal berteks (skip penetapan + target kosong)
q("""
MATCH (p:Pasal)
WHERE p.jenis IN ['substansi','ketentuan_umum','perubahan']
  AND p.text_full <> ''
SET p:Embedded
""")
```

```python
r = q("MATCH (p:Embedded) RETURN p.jenis AS j, count(*) AS c ORDER BY c DESC")
for x in r:
    print(x["j"], x["c"])
print("total Embedded:", q("MATCH (p:Embedded) RETURN count(p) AS c")[0]["c"])
```

```python
store = Neo4jVector.from_existing_graph(
    embedding=embeddings,
    url=os.environ["NEO4J_URI"],
    username=os.environ["NEO4J_USERNAME"],
    password=os.environ["NEO4J_PASSWORD"],
    index_name="pasal_embedding",
    keyword_index_name="pasal_keyword",     # fulltext index
    node_label="Embedded",
    text_node_properties=["text_full"],
    embedding_node_property="embedding",
    search_type="hybrid",                   # vector + fulltext
)
print("selesai")
```

NEO4J RETRIEVER

```python
query_txt = "regulasi pekerja kontrak"
```

```python
def fmt(record):
    md = {
        "pasal_id": record.get("pasal_id"),
        "uu": record.get("uu"),
        "score": record.get("score"),
        "diubah": [d for d in (record.get("diubah") or []) if d],
        "rujukan": record.get("rujukan") or [],
    }
    return RetrieverResultItem(content=record.get("teks"), metadata=md)
```

```python
## Hybrid + Graph (neo4j-graphrag)

retriever = HybridCypherRetriever(
    driver,
    "pasal_embedding",
    "pasal_keyword",
    retrieval_query="""
        WITH node AS anchor, score
        OPTIONAL MATCH (anchor)-[:BAGIAN_DARI]->(u:UU)
        OPTIONAL MATCH (anchor)<-[e:MENGUBAH|MENGHAPUS|MENYISIPKAN]-(ubah:Pasal)
        OPTIONAL MATCH (anchor)-[:MERUJUK]->(rujuk:Pasal)
        RETURN anchor.text_full AS teks,
            anchor.pasal_id AS pasal_id,
            score,
            coalesce(u.judul,'') AS uu,
            collect(DISTINCT {
                aksi: e.aksi,
                teks: e.teks_baru,
                dari: ubah.pasal_id,
                target: e.pasal_target
            }) AS diubah,
            collect(DISTINCT rujuk.pasal_id)[0..3] AS rujukan
    """,
    embedder=embeddings,
    result_formatter=fmt,
)
```

```python
retriever_result = retriever.search(query_text=query_txt, top_k=3)
```

```python
for item in retriever_result.items:
    print("=" * 50)
    print(item.metadata["pasal_id"], "| UU:", item.metadata["uu"], "| score:", item.metadata["score"])
    print(item.content[:300])
    print("  diubah :", item.metadata["diubah"])
    print("  rujukan:", item.metadata["rujukan"])
```

```python
## Vector + Graph (neo4j-graphrag)

vector_cypher_retriever = VectorCypherRetriever(
    driver,
    index_name="pasal_embedding",
    retrieval_query="""
        WITH node AS anchor, score
        OPTIONAL MATCH (anchor)-[:BAGIAN_DARI]->(u:UU)
        OPTIONAL MATCH (anchor)<-[e:MENGUBAH|MENGHAPUS|MENYISIPKAN]-(ubah:Pasal)
        OPTIONAL MATCH (anchor)-[:MERUJUK]->(rujuk:Pasal)
        RETURN anchor.text_full AS teks,
            anchor.pasal_id AS pasal_id,
            score,
            coalesce(u.judul,'') AS uu,
            collect(DISTINCT {
                aksi: e.aksi,
                teks: e.teks_baru,
                dari: ubah.pasal_id,
                target: e.pasal_target
            }) AS diubah,
            collect(DISTINCT rujuk.pasal_id)[0..3] AS rujukan
    """,
    embedder=embeddings,
    result_formatter=fmt
)
```

```python
hasil_cypher_retriever = vector_cypher_retriever.search(query_text="regulasi pekerja kontrak", top_k=3)
```

```python
for i, item in enumerate(hasil_cypher_retriever.items, 1):
    md = item.metadata
    print("=" * 60)
    print(f"    skor: {md['score']:.3f}")
    print("-" * 60)
    print((item.content or "")[:300])

    if md["diubah"]:
        print("\n  >> DIUBAH OLEH CIPTAKER:")
        for d in md["diubah"]:
            print(f"     dari {d['dari']} (target pasal {d['target']}, {d['aksi']})")
    if md["rujukan"]:
        print("  >> merujuk:", md["rujukan"])
    print()
```
